```python
from sklearn import datasets, linear_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, accuracy_score
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from math import ceil

import warnings

warnings.filterwarnings("ignore")
```

```python
data = datasets.load_diabetes()
```

```python
print(data.DESCR)
```

```python
data
```

```python
df = pd.DataFrame(data=data.data, columns=data.feature_names)
df["target"] = data.target
df.head()
```

```python
corr = df.corr()["target"].sort_values(ascending=False)
corr
```

```python
t_cols = len(data.feature_names)
fig,ax = plt.subplots(ceil(t_cols/3),3, figsize=(20,20))

row_i = 0
col_i = 0
for i,col in enumerate(data.feature_names):
    sns.lineplot(data=df, x=col, y="target", ax=ax[row_i, col_i])
    col_i += 1
    if(col_i%3==0):
        row_i += 1
        col_i = 0
fig.delaxes(ax[-1,-1])
fig.delaxes(ax[-1,-2])
plt.show()
```

```python
better_features = [["bmi", "s5"], ["bmi", "s6"]]
```

```python
target = "target"
```

```python
samples = 1000
```

```python
data = {}
```

```python
for features in better_features:
    features_label = " ".join(features)
    data[features_label] = []

    for _ in range(samples):
        X_train, X_test, y_train, y_test = train_test_split(df[features],df[target],test_size=0.3)
        model = linear_model.LinearRegression()
        model.fit(X_train,y_train)
        pred = model.predict(X_test)
        data[features_label].append(r2_score(y_test, pred))
```

```python
samples_results = pd.DataFrame(data)
samples_results.head()
```

```python
sns.boxplot(data)
plt.show()
```
