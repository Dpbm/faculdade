```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# para ficar bonito na hora de visualizar a tabela
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

# formata os valores float para duas casas decimais
pd.options.display.float_format = "{:.2f}".format
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
valores_unicos = []
for i in df.columns[:12].tolist():
    counts = len(df[i].astype(str).value_counts())
    print(i,':',counts)
    valores_unicos.append(counts)
```

```python
df.describe()
```

```python
df.groupby(["Sex"]).size()
```

```python
df.Sex.value_counts().plot(kind="bar", title="Sexo", color=["magenta", "cyan"])
plt.show()
```

```python
for column in df.columns:
    plt.figure(figsize=(15,10))
    df[column].value_counts().plot(kind="bar", title=column,color=sns.color_palette("rocket"))
    plt.show()
```

```python
for column in df.columns:
    print(df.groupby([column]).size())
```
