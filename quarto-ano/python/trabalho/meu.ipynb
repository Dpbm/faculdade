```python
import pandas as pd
```

```python
data = pd.read_csv("Superstore.csv", encoding="latin")
data.head()
```

## 1. Qual o total de lucro obtido?

```python
data.columns
```

```python
# o lucro total é dado pela soma da coluna  'Profit' do inglês 'lucro'
data.Profit.sum()
```

## 2. Quais os 5 estados com maior faturamento?

```python
data.groupby("State")["Sales"].sum().sort_values(ascending=False).head()
```

## 3. Quais os 5 estados com maior prejuízo?

```python
data.groupby("State")["Profit"].sum().sort_values().head()
```

## 4. Quantos pedidos foram feitos por tipo de envio?

```python
data["Ship Mode"].value_counts()
```

## 5. Qual o ticket médio (valor médio de vendas por pedido)?

```python
data["Sales"].mean()
```

## 6. Quantos clientes únicos existem?

```python
data["Customer ID"].nunique()
```

## 7. Qual o cliente que mais comprou?

```python
user_data = data.groupby("Customer ID")["Sales"].sum().sort_values(ascending=False).head(1)
user_data = pd.DataFrame(user_data.reset_index())
user_data["Name"] = data.loc[data["Customer ID"] == user_data["Customer ID"].loc[0]]["Customer Name"].unique()[0]
user_data
```

## 8. Quais os produtos mais vendidos (por quantidade)?

```python
data.groupby("Product Name")["Quantity"].sum().sort_values(ascending=False).head()
```

## 9. Produtos com prejuízo total (lucro negativo)

```python
products = data.groupby("Product Name")
s = pd.DataFrame(products["Profit"].sum()).reset_index()
s.loc[s["Profit"] < 0].sort_values(by="Profit", ascending=True)
```

## 10. Qual a correlação entre vendas e lucro?

```python
data[["Sales", "Profit"]].corr()
```

## 11. Vendas acima de R$ 1.000

```python
data[data["Sales"] > 1000]
```

## 12. Número de pedidos por ano

```python
data["Year"] = pd.to_datetime(data["Order Date"]).dt.year
data.groupby("Year")["Order ID"].count()
```
