As colunas desse dataset são:
- Passenger ID: ID do passageiro (número único para cada um dos passageiros)
- Survived: sobrevivente (0 = Não, 1 = Sim)
- Pclass: Classe da passagem (1 = primeira classe, 2 = segunda classe, 3 = terceira classe)
- Name: nome do passageiro
- Sex: Gênero do passageiro
- Age: Idade (em anos) do passageiro
- SibSp: número de irmãos / cônjuges a bordo do Titanic
- Parch: número de pais / filhos a bordo do Titanic
- Ticket: número do ticket
- Fare: tarifa da passagem
- Cabin: número da cabine
- Embarked: porto de embarque (C = Cherbourg, Q = Queenstown, S = Southampton)

**A cardinalidade nos ajuda a saber a quantidade de dados distintos em uma coluna**
- Se tivermos muitos valores distintos, provavelmente aquela coluna não será uma boa opção para usarmos no modelo
- Matematicamente, cardinalidade é o número de elementos de um conjunto
- Podemos verificar a cardinalidade usando o `.nunique()`

```python
data.nunique()
```

### Visualizando de forma gráfica
- Para visualizar essas informações de maneira gráfica, podemos utilizar o matplotlib
    - https://matplotlib.org/

```python
# Importando o matplotlib
import matplotlib.pyplot as plt
```

```python
# Verificando o histograma das tarifas
sns.histplot(data=data, x="fare")
plt.show()
```

```python
# Verificando o histograma das tarifas apenas para tarifas menores que 100 reais
subset = data.loc[data.fare < 100]
sns.histplot(data=subset, x="fare")
plt.show()
```

```python
# Verificando o boxplot para a coluna Fare
sns.boxplot(data=data, x="fare")
plt.show()
```

**Dependendo do visual, outras bibliotecas já podem ter opções mais prontas para usarmos, como o caso do pairplot no seaborn**
- Como cientistas, devemos escolher a ferramenta que melhor resolve o nosso problema
- O pairplot no seaborn:
    - https://seaborn.pydata.org/generated/seaborn.pairplot.html

```python
# Importando o seaborn
import seaborn as sns
```

```python
# Criando o pairplot
sns.pairplot(data=data)
plt.show()
```

```python
# Criando uma matriz de correlação entre as variáveis
correlacao = data.corr(numeric_only=True)
correlacao
```

```python
# Utilizando o heatmap do seaborn para tornar essa matriz mais visual
sns.heatmap(correlacao, annot=True)
plt.show()
```

```python
sns.lmplot(data=data, x="age", y="fare", hue="sex", markers=["o", "x"])
plt.show()
```

```python
sns.kdeplot(data=data, x="fare", hue="sex")
sns.rugplot(data=data, x="fare", hue="sex", height=.1)
plt.show()
```

```python
fig,ax = plt.subplots(1,3,figsize=(15,5))

sns.countplot(data=data, x="survived", hue="sex",ax=ax[0])
sns.countplot(data=data, x="survived", hue="alone",ax=ax[1])
sns.countplot(data=data, x="survived", hue="adult_male",ax=ax[2])
plt.show()
```

```python
sns.relplot(data=data, x="parch", y="age", hue="survived")
plt.show()
```

```python
fig,ax = plt.subplots(1,2,figsize=(10,5))
sns.countplot(data=data, x="pclass", hue="survived", ax=ax[0])
sns.countplot(data=data, x="pclass", hue="sex", ax=ax[1])
plt.show()
```

```python
sns.displot(data=data, x="fare", col="sex")
plt.show()
```

```python
sns.displot(data=data, x="fare", col="pclass")
plt.show()
```

```python
fig,ax = plt.subplots(1,2,figsize=(10,5))
sns.barplot(data=data, x="parch", y="fare", ax=ax[0])
sns.barplot(data=data, x="sibsp", y="fare", ax=ax[1])
plt.show()
```

```python
clean = data.loc[data.fare < 200]
sns.boxplot(data=clean, x="fare", hue="survived")
plt.show()
```

```python
sns.lineplot(data=clean, y="fare", x="parch")
sns.lineplot(data=clean, y="fare", x="sibsp")
plt.xlabel("relatives")
plt.show()
```
