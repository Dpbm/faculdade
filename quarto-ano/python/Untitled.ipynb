```python
from datetime import date
cadastral["data_atual"] = date.today()
```

```python
cadastral.head()
```

```python
cadastral.dtypes
```

```python
cadastral["data_atual"] = cadastral["data_atual"].astype('datetime64[ns]')
cadastral.head()
```

```python
cadastral.dtypes
```

```python
cadastral["idade"] = (cadastral["data_atual"] - cadastral["datanasc"]) / 365.25
cadastral.head()
```

```python
cadastral.dtypes
```

```python
import numpy as np
cadastral["idade"] = cadastral["idade"].dt.days.astype(int)
cadastral.head()
```

```python
cadastral.dtypes
```

```python
cadastral.describe().idade
```

```python
cadastral["sexo_organizado"] = pd.cut(cadastral["sexo"], bins=[0,1,2], labels=["Feminino", "Masculino"])
cadastral.head()
```

```python
cadastral.sexo_organizado.value_counts()
```

```python
cadastral["Estado_Civil"] =  pd.cut(cadastral["estado_civil"], bins=[0,1,2], labels=["casado", "solteiro"])
cadastral["plano_pagamento"] =  pd.cut(cadastral["pagamento_plano"], bins=[0,1,2,3], labels=["debito conta", "boleto", "cartao credito"])
cadastral["cat_emprego"] =  pd.cut(cadastral["catemp"], bins=[0,1,2,3], labels=["A", "B", "C"])
cadastral["faixa_etaria"] =  pd.cut(cadastral["idade"], bins=[min(cadastral.idade)-1,58,62,76,max(cadastral.idade)+1], labels=["A", "B", "C", "D"])
```

```python
print(cadastral["Estado_Civil"].value_counts())
print(cadastral["plano_pagamento"].value_counts())
print(cadastral["cat_emprego"].value_counts())
print(cadastral["faixa_etaria"].value_counts())
```
