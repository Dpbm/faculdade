```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# para ficar bonito na hora de visualizar a tabela
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

# formata os valores float para duas casas decimais
pd.options.display.float_format = "{:.2f}".format
```

```python
variaveis_numericas = []
for i in df.columns[:11].tolist():
    if df.dtypes[i] == 'int64' or df.dtypes[i] == 'float64':
        variaveis_numericas.append(i)
```

```python
variaveis_numericas
```

```python
plt.rcParams["figure.figsize"]=[14,20]
plt.rcParams["figure.autolayout"]=True

f,ax = plt.subplots(3,2)
linha = 0
coluna = 0

                
for i in variaveis_numericas:
    sns.boxplot(data=df,y=i,ax=ax[linha,coluna])
    coluna += 1
    if coluna == 2:
        linha += 1
        coluna = 0
```

```python
plt.rcParams["figure.figsize"]=[14,20]
plt.rcParams["figure.autolayout"]=True

f,ax = plt.subplots(3,2)
linha = 0
coluna = 0

                
for i in variaveis_numericas:
    sns.histplot(data=df,x=i,ax=ax[linha,coluna])
    coluna += 1
    if coluna == 2:
        linha += 1
        coluna = 0
```
