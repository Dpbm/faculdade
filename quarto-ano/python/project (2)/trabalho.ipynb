```python
%matplotlib inline
import os
import gc

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
BASE_DATA_FOLDER = os.path.join(".", "base_data")
```

```python
def read_csv(file:str) -> pd.DataFrame:
    ENCODING="windows-1252"
    SEP=";"
    return pd.read_csv(file, encoding=ENCODING, sep=SEP)
```

```python
def concat_path(file:str) -> str:
    return os.path.join(BASE_DATA_FOLDER, file)
```

```python
csv_files = list(filter(lambda file: 'csv' in file, [concat_path(file)  for file in os.listdir(path=BASE_DATA_FOLDER)]))
csv_files
```
