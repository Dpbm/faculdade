```python
from sklearn import datasets
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = sns.load_dataset('penguins')
data = data.dropna()

data.head()
```

```python
target = "species"
features = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
```

```python
n_elements = len(features)
counter = 1

plt.figure(figsize=(20,20))

for i in features:
    for j in features:
        plt.subplot(n_elements,n_elements,counter)
        sns.scatterplot(data=data, x=i,y=j,hue=target)
        counter += 1
plt.show()
```

```python
selected_features = ["bill_length_mm", "flipper_length_mm"]
```

```python
model = KMeans(n_clusters=3, init="random")
```

```python
X = data[selected_features]
y = target
```

```python
model.fit(X)
```

```python
print(model.cluster_centers_)
```

```python
distancies = model.fit_transform(X)
```

```python
model.labels_
```

```python
plt.scatter(X[selected_features[0]], X[selected_features[1]], s=50,  c=model.labels_)
plt.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1], s=200, c="red", label="centroides")
plt.show()
```
