```python
df = df.sort_values(by=["Data Venda", "ValorVenda"], ascending=False)
df.head()
```

```python
plt.figure(figsize=(20,6))
plt.plot(df["Data Venda"],df["ValorVenda"])
plt.show()
```

```python
df.columns
```

```python
df.dtypes
```

```python
df["Ano"] = df["Data Venda"].dt.year
df["Mes"] = df["Data Venda"].dt.month
df["Ano_"] = df["Ano"].astype(str)
df["Mes_"] = df["Mes"].astype(str)
df["Mes_Ano"] = df["Mes_"] + "_" + df["Ano_"]
df.head()
```

```python
pie = df[["Mes_Ano", "ValorVenda", "ValorVendaMinima", "ValorCompra"]].groupby(["Mes_Ano"]).mean()
pie["Mes_Ano"] = pie.index
pie
```

```python
plt.figure(figsize=(40,10))
plt.plot(pie["Mes_Ano"], pie["ValorVenda"])
plt.xticks(rotation=90)
plt.show()
```

```python
pie_ano = df[["Ano", "ValorVenda", "ValorVendaMinima", "ValorCompra"]].groupby(["Ano"]).mean()
pie_ano["Ano"] = pie_ano.index
pie_ano
```

```python
pie_mes = df[["Mes", "ValorVenda", "ValorVendaMinima", "ValorCompra"]].groupby(["Mes"]).mean()
pie_mes["Mes"] = pie_mes.index
pie_mes
```

```python
pie_mes.columns, pie_mes.dtypes
```

```python
plt.figure(figsize=(20,5))
plt.xticks(pie_mes["Mes"])
plt.ylim(min(pie_mes.ValorVenda), max(pie_mes.ValorVenda))
plt.plot(pie_mes["Mes"], pie_mes["ValorVenda"])
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(1,2,sharex=True,sharey=True)
pricemin = pie_mes["ValorVenda"].min()
plt.xticks(pie_mes["Mes"])
ax1.plot(pie_mes["Mes"], pie_mes["ValorVenda"], lw=3)
ax2.fill_between(pie_mes["Mes"], pricemin, pie_mes["ValorVenda"], alpha=0.9)

for ax in (ax1,ax2):
    ax.grid(True)
    ax.label_outer()
ax1.set_ylabel("Preco")
fig.suptitle("Menor preco de vendas no periodo")
fig.autofmt_xdate()
```

```python
barras = df[["Vendedor", "ID-Produto"]].groupby(["Vendedor"]).count()
barras["Vendedor"] = barras.index
barras.head()
```

```python
barras["Vendedor"].shape
```

```python
fig,ax=plt.subplots(figsize=(20,7))
bar_colors = [
    "tab:red",
    "tab:blue",
    "tab:green",
    "tab:pink",
    "tab:orange",
    "tab:pink",
    "tab:red",
    "tab:blue",
    "tab:orange",
]
vendedor = barras.Vendedor
qt_vendas = barras["ID-Produto"]
ax.bar(vendedor, qt_vendas,label=vendedor,color=bar_colors)
ax.set_ylabel("Qtdade Vendas")
ax.set_title("Vendas por Funcionarios")
ax.legend(title="Funcionarios")
plt.show()
```
