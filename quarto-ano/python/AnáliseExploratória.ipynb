## 5. Crie histogramas para 'Income' e 'Total Population'.

```python
plt.figure(figsize=(16, 6))

plt.subplot(1, 2, 1)
sns.histplot(df_census['Income'].dropna(), kde=True, bins=40, color='skyblue')
plt.title('Distribuição de Renda Média por Setor Censitário')
plt.xlabel('Renda Média (Income)')
plt.ylabel('Frequência')

plt.subplot(1, 2, 2)
sns.histplot(df_census['TotalPop'].dropna(), kde=True, bins=40, color='salmon')
plt.title('Distribuição da População Total por Setor Censitário')
plt.xlabel('População Total (TotalPop)')
plt.ylabel('Frequência')

plt.suptitle('Histogramas de Renda e População Total', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

##  6. Crie boxplots se necessário para visualizar dispersões.

```python
plt.figure(figsize=(16, 6))

plt.subplot(1, 2, 1)
sns.boxplot(y=df_census['Income'].dropna(), color='skyblue')
plt.title('Boxplot da Renda Média')
plt.ylabel('Renda Média (Income)')

plt.subplot(1, 2, 2)
sns.boxplot(y=df_census['Poverty'].dropna(), color='lightcoral')
plt.title('Boxplot da Taxa de Pobreza')
plt.ylabel('Taxa de Pobreza (%)')

plt.suptitle('Visualização de Dispersão para Renda e Pobreza', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

## 7. Crie um heatmap entre: 'Income', 'Employed', 'Unemployment', 'Poverty'

```python
print("\n\nHeatmap de Correlação\n")
correlation_cols = ['Income', 'Employed', 'Unemployment', 'Poverty']
correlation_matrix = df_census[correlation_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Mapa de Calor da Correlação entre Variáveis Socioeconômicas', fontsize=15)
plt.show()
```

## Pergunta 1: Quais regiões têm os maiores e menores níveis de renda?

```python
df_filtered = df_census.dropna(subset=['Income', 'TotalPop'])
borough_income = df_filtered.groupby('Borough').apply(
        lambda x: (x['Income'] * x['TotalPop']).sum() / x['TotalPop'].sum()
    ).sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sns.barplot(x=borough_income.index, y=borough_income.values, palette='viridis', hue=borough_income.index, dodge=False, legend=False)
plt.title('Média de Renda por Distrito (Borough)', fontsize=16)
plt.xlabel('Distrito (Borough)')
plt.ylabel('Renda Média Ponderada pela População ($)')
plt.tight_layout()
plt.show()
```

### O gráfico de barras mostra claramente que Manhattan tem o maior nível de renda, enquanto o Bronx tem o menor.

## Pergunta 2: Existe correlação entre pobreza e desemprego? 

```python
plt.figure(figsize=(12, 7))
sns.regplot(data=df_census, x='Unemployment', y='Poverty', scatter_kws={'alpha':0.2, 'color':'red'}, line_kws={'color':'blue'})
plt.title('Relação entre Taxa de Desemprego e Taxa de Pobreza', fontsize=16)
plt.xlabel('Taxa de Desemprego (%)')
plt.ylabel('Taxa de Pobreza (%)')
plt.tight_layout()
plt.show()
```

### Sim, os dois problemas estão claramente interligados. Nas regiões onde a taxa de desemprego é mais alta também tendem a apresentar uma taxa de pobreza mais elevada. Conforme o gráfico de dispersão a linha de regressão tem uma tendência positiva: à medida que o desemprego aumenta, a pobreza também tende a aumentar.

## Pergunta 3: Há regiões com alta renda e também alto desemprego?

```python
plt.figure(figsize=(12, 7))
sns.scatterplot(data=df_census, x='Income', y='Unemployment', alpha=0.3)
plt.title('Relação entre Renda Média e Taxa de Desemprego', fontsize=16)
plt.xlabel('Renda Média ($)')
plt.ylabel('Taxa de Desemprego (%)')
plt.tight_layout()
plt.show()
```

### Não, regiões com renda média mais alta tendem a ter taxas de desemprego mais baixas. O gráfico mostra que os pontos se concentram no canto inferior esquerdo e se espalham para a direita, indicando que regiões de alta renda (direita) geralmente têm baixo desemprego (embaixo). Não há concentração de pontos no canto superior direito (alta renda e alto desemprego)

## Pergunta 4: Qual fator parece mais influenciar a pobreza?

```python

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

sns.regplot(data=df_census, x='Income', y='Poverty', ax=ax1, scatter_kws={'alpha':0.2}, line_kws={'color':'green'})
ax1.set_title('Influência da Renda na Pobreza', fontsize=14)
ax1.set_xlabel('Renda Média ($)')
ax1.set_ylabel('Taxa de Pobreza (%)')

sns.regplot(data=df_census, x='Unemployment', y='Poverty', ax=ax2, scatter_kws={'alpha':0.2, 'color':'red'}, line_kws={'color':'blue'})
ax2.set_title('Influência do Desemprego na Pobreza', fontsize=14)
ax2.set_xlabel('Taxa de Desemprego (%)')
ax2.set_ylabel('Taxa de Pobreza (%)')

fig.suptitle('Comparando Fatores que Influenciam a Pobreza', fontsize=18)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### A falta de renda é o principal indicador associado à pobreza, mais forte até que a correlação com o desemprego.

## Pergunta 5: Onde a prefeitura deveria investir com mais prioridade?

```python

avg_poverty_borough = df_census.groupby('Borough')['Poverty'].mean().sort_values(ascending=False)
avg_unemployment_borough = df_census.groupby('Borough')['Unemployment'].mean().sort_values(ascending=False)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

sns.barplot(x=avg_poverty_borough.index, y=avg_poverty_borough.values, palette='coolwarm', ax=ax1, hue=avg_poverty_borough.index, dodge=False, legend=False)
ax1.set_title('Taxa Média de Pobreza por Distrito', fontsize=14)
ax1.set_xlabel('Distrito (Borough)')
ax1.set_ylabel('Taxa Média de Pobreza (%)')

sns.barplot(x=avg_unemployment_borough.index, y=avg_unemployment_borough.values, palette='coolwarm_r', ax=ax2, hue=avg_unemployment_borough.index, dodge=False, legend=False)
ax2.set_title('Taxa Média de Desemprego por Distrito', fontsize=14)
ax2.set_xlabel('Distrito (Borough)')
ax2.set_ylabel('Taxa Média de Desemprego (%)')

fig.suptitle('Indicadores para Prioridade de Investimento', fontsize=18)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### A prefeitura deveria priorizar investimentos nas regiões com os maiores índices combinados de pobreza e desemprego. Portanto, os investimentos deveriam ser direcionados com prioridade para o Bronx e outras áreas pontuais do Brooklyn que também aparecem nos rankings.
