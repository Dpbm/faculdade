## 5. Crie histogramas para 'Income' e 'Total Population'.

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
sns.histplot(data=df, x="Income")
plt.subplot(1,2,2)
sns.histplot(data=df, x="TotalPop")
plt.show()
```

##  6. Crie boxplots se necessário para visualizar dispersões.

```python
cols_per_row = 4
cols = df.columns
n_rows = int(np.ceil(len(cols)/cols_per_row))

plt.figure(figsize=(40,40))
for i,col in enumerate(cols):
    plt.subplot(n_rows, cols_per_row, i+1) 
    sns.boxplot(data=df, y=col)
```

## 9. Liste as 10 com maior pobreza ou desemprego.

```python
df.groupby(["Borough"])["Poverty"].mean().sort_values(ascending=False)
```

```python
df.groupby(["Borough"])["Unemployment"].mean().sort_values(ascending=False)
```

### Existe correlação entre pobreza e desemprego? 

```python
grouped = df.groupby(["Borough"])
neighborhood = df["Borough"].unique()
plt.bar(x=neighborhood,  height=grouped["Poverty"].mean(), label="Pobreza")
plt.bar(x=neighborhood,  height=grouped["Unemployment"].mean(), label="Desemprego")
plt.legend()
plt.show()
```

```python
sns.histplot(data=df,x="Poverty", label="Pobreza", binwidth=2)
sns.histplot(data=df,x="Unemployment", label="Desemprego", binwidth=2)
plt.legend()
plt.xlabel("")
plt.show()
```

```python
sns.lmplot(data=df, y="Poverty", x="Unemployment")
```

### Sim, regiões onde a pobreza é grande o desemprego
