```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
dataset = "https://raw.githubusercontent.com/yashuv/CodeClause_Market_Basket_Analysis_in_Python_using_Apriori_Algorithm/refs/heads/main/data/Groceries_dataset.csv"
```

```python
data = pd.read_csv(dataset)
data.head()
```

```python
data.drop("Date",axis=1,inplace=True)
```

```python
data["itemDescription"].nunique()
```

```python
data[data["Member_number"] == 1808].groupby(["Member_number","itemDescription"])["itemDescription"].count()
```

```python
all_data_grouped = data.groupby(["Member_number","itemDescription"])["itemDescription"].count().unstack().reset_index().fillna(0).set_index("Member_number")
all_data_grouped.head()
```

```python
def fix_values(x):
    return 0 if x <= 0 else 1
```

```python
all_data_grouped = all_data_grouped.map(fix_values)
```

```python
algo = apriori(all_data_grouped, min_support=0.06, use_colnames=True)
rules = association_rules(algo, metric="lift", min_threshold=1)
rules.sort_values(["lift", "support"]).head(10)
```
