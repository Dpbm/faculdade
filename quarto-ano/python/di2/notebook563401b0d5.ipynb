```python
transactions = [
   ['beer', 'wine', 'cheese'],
   ['beer', 'potato chips'],
   ['eggs', 'flower', 'butter', 'cheese'],
   ['eggs', 'flower', 'butter', 'beer', 'potato chips'],
   ["wine", 'cheese'],
   ['potato chips'],
   ['eggs', 'flower', 'butter', 'wine', 'cheese'],
   ['eggs', 'flower', 'butter', 'beer', 'potato chips'],
   ['wine', 'beer'],
   ['beer', 'potato chips'],
   ['butter', 'eggs'],
   ['beer', 'potato chips'],
   ['flower', 'eggs'],
   ['beer', 'potato chips'],
   ['eggs', 'flower', 'butter', 'wine', 'cheese'],
   ['beer', 'wine', 'potato chips', 'cheese'],
   ['wine', 'cheese'],
   ['beer', 'potato chips'],
   ['wine', 'cheese'],
   ['beer', 'potato chips']
]
```

```python
import pandas as pd

data = pd.DataFrame(transactions)
data
```

```python
min_n_products = 2
min_support = 7/len(transactions)
max_length = max([len(x) for x in transactions])
```

```python
from pyECLAT import ECLAT
my_eclat = ECLAT(data=data, verbose=True)
rules_indices, rule_supports = my_eclat.fit(min_support=min_support, min_combination=min_n_products, max_combination=max_length)
```
