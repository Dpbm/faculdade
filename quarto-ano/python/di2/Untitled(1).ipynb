```python
from datetime import date
cadastral["data_atual"] = date.today()
```

```python
cadastral.head()
```

```python
cadastral.dtypes
```

```python
cadastral["data_atual"] = cadastral["data_atual"].astype('datetime64[ns]')
cadastral.head()
```

```python
cadastral.dtypes
```

```python
cadastral["idade"] = (cadastral["data_atual"] - cadastral["datanasc"]) / 365.25
cadastral.head()
```

```python
cadastral.dtypes
```

```python
import numpy as np
cadastral["idade"] = cadastral["idade"].dt.days.astype(int)
cadastral.head()
```

```python
cadastral.dtypes
```

```python
cadastral.describe().idade
```

```python
cadastral["sexo_organizado"] = pd.cut(cadastral["sexo"], bins=[0,1,2], labels=["Feminino", "Masculino"])
cadastral.head()
```

```python
cadastral.sexo_organizado.value_counts()
```

```python
cadastral["Estado_Civil"] =  pd.cut(cadastral["estado_civil"], bins=[0,1,2], labels=["casado", "solteiro"])
cadastral["plano_pagamento"] =  pd.cut(cadastral["pagamento_plano"], bins=[0,1,2,3], labels=["debito conta", "boleto", "cartao credito"])
cadastral["cat_emprego"] =  pd.cut(cadastral["catemp"], bins=[0,1,2,3], labels=["A", "B", "C"])
cadastral["faixa_etaria"] =  pd.cut(cadastral["idade"], bins=[min(cadastral.idade)-1,58,62,76,max(cadastral.idade)+1], labels=["A", "B", "C", "D"])
```

```python
print(cadastral["Estado_Civil"].value_counts())
print(cadastral["plano_pagamento"].value_counts())
print(cadastral["cat_emprego"].value_counts())
print(cadastral["faixa_etaria"].value_counts())
```

```python
cadastral.head()
```

```python
cadastral["faixa_etaria"] = pd.qcut(cadastral["idade"],q=3)
cadastral.head()
```

```python
cadastral["faixa_etaria"].value_counts()
```

```python
cadastral[["sexo_organizado","renda_familiar"]].groupby(["sexo_organizado"]).mean()
```

```python
pd.pivot_table(cadastral, values="renda_familiar", index=["Estado_Civil", "sexo_organizado"], columns=["cat_emprego"], aggfunc=np.mean)
```

```python
pd.pivot_table(cadastral, values="valor_prestacao", index=["Estado_Civil"], columns=["sexo_organizado"], aggfunc=np.mean)
```

```python
cadastral["comp_renda"] = ((cadastral.valor_prestacao/cadastral.renda_familiar)*100).round(2)
cadastral.head()
```

```python
transacional = pd.read_excel("Arquivo Transacional.xlsx")
transacional.head()
```

```python
transacional["churn_categorizado"] = transacional["Churn"].map({0:"ativo",1:"churn"})
transacional.head()
```

```python
transacional.sort_values(['id']).head()
```

```python
transacional["tempo_ligacao"] = (transacional.longdist + transacional.internat + transacional.local).round(2)
transacional.head()
```

```python
consolidado = pd.merge(cadastral, transacional, on="id")
consolidado.head()
```
