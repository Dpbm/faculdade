```python
import pandas as pd
df = pd.read_csv('/content/Superstore.csv' , encoding = 'latin')
df
```

```python
df.columns
```

```python
# 1. Qual o total de lucro obtido?
df['Profit'].sum()
```

```python
# 2. Quais os 5 estados com maior faturamento?
df.groupby('State')['Sales'].sum().nlargest(5)
```

```python
# 3. Quais os 5 estados com maior prejuízo?
df.groupby('State')['Profit'].sum().sort_values().head(5)
```

```python
# 4. Quantos pedidos foram feitos por tipo de envio?
df['Ship Mode'].value_counts()
```

```python
# 5. Qual o ticket médio (valor médio de vendas por pedido)?
df['Sales'].mean()
```

```python
# 6. Quantos clientes únicos existem?
df['Customer ID'].nunique()
```

```python
# 7. Qual o cliente que mais comprou?
df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(1)
```

```python
# 8. Quais os produtos mais vendidos (por quantidade)?
df.groupby('Product Name')['Quantity'].sum().sort_values(ascending=False).head(10)
```

```python
# 9. Produtos com prejuízo total (lucro negativo)
df.groupby('Product Name')['Profit'].sum().sort_values().head(10)
```

```python
# 10. Qual a correlação entre vendas e lucro?
df[['Sales', 'Profit']].corr()
```

```python
# 11. Vendas acima de R$ 1.000
df[df['Sales'] > 1000]
```

```python
# 12. Número de pedidos por ano
df['Year'] = pd.DatetimeIndex(df['Order Date']).year
df.groupby('Year')['Order ID'].count()
```
