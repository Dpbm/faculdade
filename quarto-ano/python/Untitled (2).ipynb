```python
import pandas as pd
import seaborn as sns
import summarytools as st
from summarytools import dfSummary
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
df = pd.read_excel("Banco.xlsx")
df.head()
```

```python
def contar_valores_percentual(data):
    variaveis_object = data.select_dtypes(include=["object"]).columns

    tabelas = []
    for column in variaveis_object:
        tabela_tmp = data[column].value_counts(normalize=False).reset_index()
        tabela_tmp.columns = ["Valor", "Contagem"]
        tabela_tmp["Percentual"] = (data[column].value_counts(normalize=True).values * 100).round(2)
        tabela_tmp["Variavel"]=column
        tabelas.append(tabela_tmp)
    resultado = pd.concat(tabelas, ignore_index=True)
    return resultado[["Variavel", "Valor", "Contagem", "Percentual"]]
```

```python
contar_valores_percentual(df)
```

```python
bins = [
    df["estudo"].min(),
    df["estudo"].quantile(0.2),
    df["estudo"].quantile(0.6),
    df["estudo"].quantile(0.8),
    df["estudo"].max(),
]

grupos = ["Graduacao","Mestrado","Doutorado", "Pos-Doutorado"]

df["faixa_estudo"]=pd.cut(df["estudo"],bins=bins,labels=grupos, include_lowest=True, duplicates="drop")

df["faixa_estudo"].value_counts()
    
```

```python
df["faixa_estudo"] = df["faixa_estudo"].replace({"Doutorado":"Outros", "Pos-Doutorado":"Outros"})
df["faixa_estudo"].value_counts()
```

```python
faixa_estudo_counts = df["faixa_estudo"].value_counts()
faixa_estudo_percentual = df["faixa_estudo"].value_counts(normalize=True)*100
```

```python
resultado = pd.DataFrame({
    "Contagem":faixa_estudo_counts,
    "Percentual":faixa_estudo_percentual.round(2)
})
resultado.loc["Total"] = [resultado["Contagem"].sum(), 100.0]
print(resultado)
```

```python
absoluto = pd.crosstab(df.sexo, df.faixa_estudo, margins=True, margins_name="Total")
percentual = pd.crosstab(df.sexo, df.faixa_estudo, margins=True, margins_name="Total", normalize="all")
percentual = percentual * 100
resultado = absoluto.astype(str) + " (" + percentual.round(2).astype(str) + "%)"
resultado
```
