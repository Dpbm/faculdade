```python
import os
import random
import matplotlib.pyplot as plt
import sys
```

```python
n = 1000
choices = [0, 0, 1]
default = [random.choice(choices) for _ in range(n)]
newseed = []
```

```python
for j in range(n):
    seed = int.from_bytes(os.urandom(64), byteorder=sys.byteorder, signed=False) + (j*2133)
    random.seed()
    newseed.append(random.choice(choices))
```

```python
amount_default_zeros = default.count(0)
amount_newseed_zeros = newseed.count(0)

default_data = {
    0:amount_default_zeros,
    1:n-amount_default_zeros
}
newseed_data = {
    0:amount_newseed_zeros,
    1:n-amount_newseed_zeros
}
```

```python
default_data
```

```python
fig,ax = plt.subplots(1,2)
labels = ["0", "1"]
colors = ["blue", "orange"]

ax[0].bar(x=labels, height=default_data.values(), color=colors)
ax[1].bar(x=labels, height=newseed_data.values(), color=colors)
```
