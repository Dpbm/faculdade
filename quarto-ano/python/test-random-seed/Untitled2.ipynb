```python
from sklearn import datasets
dados = datasets.load_iris()
print(dados)
```

```python
print(dados.target_names)
print(dados.target)
```

```python
print(dados.feature_names)
print(dados.data)
```

```python
import pandas as pd
total = pd.DataFrame(data=dados.data, columns=dados.feature_names)
total['target'] = dados.target
total.head()
```

```python
!pip install seaborn --break-system-packages
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plot = sns.FacetGrid(total, hue="target")
plot.map(sns.distplot, "sepal length (cm)").add_legend()

plot = sns.FacetGrid(total, hue="target")
plot.map(sns.distplot, "sepal width (cm)").add_legend()

plot = sns.FacetGrid(total, hue="target")
plot.map(sns.distplot, "petal length (cm)").add_legend()

plot = sns.FacetGrid(total, hue="target")
plot.map(sns.distplot, "petal width (cm)").add_legend()

plt.show()
```

```python
X = dados.data[:, 2:4]
print(X)
```

```python
parcial = pd.DataFrame(data=X, columns=dados.feature_names[2:4])
parcial['target'] = dados.target
parcial.head()
```

```python
parcial.plot(kind="scatter", x="petal length (cm)", y="petal width (cm)", c="target", cmap="viridis")
```

```python
sns.scatterplot(data=parcial, x="petal length (cm)", y="petal width (cm)", hue="target", palette="Set1")
```

```python
from sklearn.model_selection import train_test_split
y=dados.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn import tree
modelo = tree.DecisionTreeClassifier()
modelo.fit(X_train, y_train)
```

```python
plt.figure(figsize=(20,12))
grafico = tree.plot_tree(modelo, feature_names=dados.feature_names, class_names=dados.target_names,
                         rounded=True, filled=True, fontsize=14)
```
